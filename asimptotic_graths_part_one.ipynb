Import the necessary packages

```python
import matplotlib.pyplot as plt
import numpy as np
```

Write the answers cpp code gave in result to different N

```python
x = np.array([100, 500, 1000, 5000, 10000, 50000, 100000, 500000, 1000000])
```

```python
y_binary_search = np.array([899, 1120, 1223, 1448, 1574, 1858, 2003, 2460, 2806])
y_search = np.array([3, 16, 30, 87, 136, 577, 1104, 5530, 11122])
```

Plot them

```python
plt.scatter(x, y_binary_search)
plt.title("Binary search")
plt.xlabel("Length of the array")
plt.ylabel("Time")
```

```python
plt.scatter(x, y_search)
plt.title("Search")
plt.xlabel("Length of the array")
plt.ylabel("Time")
```

```python
plt.scatter(x, y_binary_search)
plt.scatter(x, y_search)
plt.title("Both on the same plot")
plt.xlabel("Length of the array")
plt.ylabel("Time")
```
