```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv('shell_sort_data.csv')
print(df)
```

```python
x = np.array(df.iloc[:, 0])
y1 = np.array(df.iloc[:, 1])
y2 = np.array(df.iloc[:, 2])
y3 = np.array(df.iloc[:, 3])
```

```python
plt.scatter(x, y1)
plt.xlabel('n')
plt.ylabel('Time, ms')
plt.title('Shell sort one')
```

```python
plt.scatter(x, y2)
plt.xlabel('n')
plt.ylabel('Time, ms')
plt.title('Shell sort one')
```

```python
plt.scatter(x, y3)
plt.xlabel('n')
plt.ylabel('Time, ms')
plt.title('Shell sort one')
```

```python
plt.scatter(x, y1, label="Shell sort one")
plt.scatter(x, y2, label="Shell sort two")
plt.scatter(x, y3, label="Shell sort three")

plt.xlabel('n')
plt.ylabel('Time, ms')
plt.title('Shell sort all')
plt.legend()
```

```python
plt.scatter(x, y1, label="Shell sort one")
plt.scatter(x, y2, label="Shell sort two")
plt.scatter(x, y3, label="Shell sort three")

plt.xlabel('n')
plt.ylabel('Time, ms')
plt.title('Shell sort all')
plt.legend()


def func(x, a, b):
    return a * (x **b)

popt, pcov = curve_fit(func, x, y1)
plt.plot(x, func(x, *popt), label="Shell sort one approximation", linestyle='--')
print(*popt[1:])

popt, pcov = curve_fit(func, x, y2)
plt.plot(x, func(x, *popt), label="Shell sort one approximation", linestyle='--')
print(*popt[1:])

popt, pcov = curve_fit(func, x, y3)
plt.plot(x, func(x, *popt), label="Shell sort one approximation", linestyle='--')
print(*popt[1:])
```
